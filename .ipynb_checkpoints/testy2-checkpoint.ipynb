```python
from tkinter import Tk, filedialog
from pydub import AudioSegment
```

```python
# Ukrycie głównego okna Tkinter
root = Tk()
root.withdraw()
# Ustawiamy je tymczasowo na wierzchu
root.attributes("-topmost", True)
```

```python
# Wybór pliku MP4
video_path = filedialog.askopenfilename(
    title="Wybierz plik MP4",
    filetypes=[("Pliki MP4", "*.mp4"), ("Wszystkie pliki", "*.*")]
)
```

```python
# Po wyborze wyłączamy tryb "zawsze na wierzchu"
root.attributes("-topmost", False)

# Zamykamy Tkinter
root.destroy()

# Wczytanie pliku
audio = AudioSegment.from_file(video_path)

# Zapis MP3 w tym samym folderze
audio_path = video_path.rsplit(".", 1)[0] + ".mp3"

audio.export(audio_path, format="mp3")

print(f"Plik MP3 zapisano tutaj: {audio_path}")
```
